# 모델 로딩

```python
import torch
import json
from transformers import pipeline, GenerationConfig
```

```python
torch.cuda.is_available()
```

```python
# Gemma 3 1B (텍스트 전용, Instruction-tuned) 모델
# 사용 전 huggingface.co/google/gemma-3-1b-it 에서 라이선스 동의가 필요합니다.
model_id = "google/gemma-3-1b-it"
```

```python
pipe = pipeline(
    "text-generation",
    model=model_id,
    dtype=torch.bfloat16,
    device_map="auto",
)
```

```python
pipe.generation_config.max_new_tokens = 256 # 새로 생성할 토큰 수만 지정
pipe.generation_config.max_length = None # 입력 프롬프트 + 새로 생성되는 토큰
```

```python
messages = [
    {"role": "system", "content": "You are a pirate chatbot who always responds in pirate speak!"},
    {"role": "user", "content": "Who are you?"},
]
```

```python
outputs = pipe(
    messages,
    clean_up_tokenization_spaces=False # Gemma는 BPE(Byte-Pair Encoding) 방식은 애초에 공백 자체를 토큰의 일부로 포함해서 다룸.
)
```

```python
outputs
```

```python
print(outputs[0]["generated_text"][-1]["content"])
```

```python
print(json.dumps(outputs, ensure_ascii=False, indent=4))
```
